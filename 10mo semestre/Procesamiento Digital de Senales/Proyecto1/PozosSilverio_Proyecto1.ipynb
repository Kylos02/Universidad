```python
import math

def bit_reversal(n, bits):
    binario = f"{n:0{bits}b}"
    return int(binario[::-1], 2)

def generar_fft_svg(nombre_archivo="fft_1024_etiquetada.svg", N=1024):
    etapas = int(math.log2(N))

    # Configuración de dimensiones del lienzo
    ancho = 1200
    alto = 20000
    margen_x = 250 # Margen ampliado para que quepan las etiquetas de texto
    margen_y = 100

    dx = ancho / etapas
    dy = alto / N

    print(f"Generando diagrama FFT radix-2 DIT para N={N} con etiquetas...")

    with open(nombre_archivo, "w") as f:
        # Cabecera del archivo SVG
        f.write(f'<svg xmlns="http://www.w3.org/2000/svg" width="{ancho + 2*margen_x}" height="{alto + 2*margen_y}">\n')
        f.write('<rect width="100%" height="100%" fill="#ffffff"/>\n')

        # 1. Dibujar las conexiones de las Mariposas (Lógica DIT)
        for s in range(etapas):
            distancia = 2**s
            for bloque in range(0, N, 2 * distancia):
                for k in range(distancia):
                    idx_superior = bloque + k
                    idx_inferior = bloque + k + distancia

                    x1 = margen_x + s * dx
                    x2 = margen_x + (s + 1) * dx
                    y_sup = margen_y + idx_superior * dy
                    y_inf = margen_y + idx_inferior * dy

                    # Líneas horizontales
                    f.write(f'<line x1="{x1}" y1="{y_sup}" x2="{x2}" y2="{y_sup}" stroke="#343a40" stroke-width="0.6" opacity="1.0"/>\n')
                    f.write(f'<line x1="{x1}" y1="{y_inf}" x2="{x2}" y2="{y_inf}" stroke="#343a40" stroke-width="0.6" opacity="1.0"/>\n')

                    # Líneas cruzadas
                    f.write(f'<line x1="{x1}" y1="{y_sup}" x2="{x2}" y2="{y_inf}" stroke="#0d6efd" stroke-width="0.6" opacity="1.0"/>\n')
                    f.write(f'<line x1="{x1}" y1="{y_inf}" x2="{x2}" y2="{y_sup}" stroke="#dc3545" stroke-width="0.6" opacity="1.0"/>\n')

        # 2. Dibujar Nodos y Etiquetas
        for s in range(etapas + 1):
            x = margen_x + s * dx

            # Títulos de las columnas (Etapas)
            if s == 0:
                f.write(f'<text x="{x}" y="{margen_y - 25}" font-family="Computer Modern" font-size="16" font-weight="bold" text-anchor="middle">Entrada (Bit-Reversal)</text>\n')
            elif s == etapas:
                f.write(f'<text x="{x}" y="{margen_y - 25}" font-family="Computer Modern" font-size="16" font-weight="bold" text-anchor="middle">Salida (Orden Natural)</text>\n')
            else:
                f.write(f'<text x="{x}" y="{margen_y - 25}" font-family="Computer Modern" font-size="14" font-weight="bold" text-anchor="middle" fill="#495057">Etapa {s}</text>\n')

            for i in range(N):
                y = margen_y + i * dy

                # Dibujar nodo interactivo (Tooltip al pasar el ratón)
                f.write('<g>\n')
                f.write(f'  <title>Etapa {s}, Fila {i}</title>\n')
                f.write(f'  <circle cx="{x}" cy="{y}" r="2" fill="#212529" style="cursor: pointer;"/>\n')
                f.write('</g>\n')

                # Textos visibles explícitos en Entrada y Salida
                if s == 0:
                    idx_bitrev = bit_reversal(i, etapas)
                    f.write(f'<text x="{x - 12}" y="{y + 4}" font-family="monospace" font-size="10" text-anchor="end" fill="#000">x[{idx_bitrev}]</text>\n')
                elif s == etapas:
                    f.write(f'<text x="{x + 12}" y="{y + 4}" font-family="monospace" font-size="10" text-anchor="start" fill="#000">X[{i}]</text>\n')

        f.write('</svg>\n')

    print(f"¡Éxito! Archivo guardado como: {nombre_archivo}")

if __name__ == "__main__":
    # Prueba primero con N=16 para ver cómo queda, luego cámbialo a 1024
    generar_fft_svg(N=1024)
```
